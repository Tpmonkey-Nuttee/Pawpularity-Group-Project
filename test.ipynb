```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


df = pd.read_csv("./dataset/train.csv")
pawp = df["Pawpularity"].values.squeeze()

pawp
```

```python
plt.hist(pawp)
plt.show()
```

```python
n = np.log1p(pawp) / np.log1p(pawp.max())
plt.hist(n)
plt.show()
```

```python
r = np.expm1(n * np.log1p(pawp.max()))
plt.hist(r)
plt.show()
```
